**Bag of Words**
This is a technique used in natural language processing to represent a text document as a numerical vector.Its basic idea is to convert a Word document to a vector of word counts, where each dimension of the vector corresponds to a speciifi word in the vocabulary.

Term Frequency and Inverse Document Frequency (TF-IDF)
It is a statistical measure that is used to evaluate the importance of a word in a document or a corpus of documents.

W(x,y) = tf(x,y) * log (N/df(x))
N is number of documents
